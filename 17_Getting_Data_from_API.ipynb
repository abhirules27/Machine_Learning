**API - Application Programming Interface**

```python
# Getting data from TMDB Website
# Go to JSON Viewer Website
# API ab9e733b96cf706676a9861f00bbab20
# Using website http://api.themoviedb.org/3/movie/top_rated?api_key=ab9e733b96cf706676a9861f00bbab20&&language=en-US&page=1
```

```python
import pandas as pd
import requests
```

```python
response = requests.get('http://api.themoviedb.org/3/movie/top_rated?api_key=ab9e733b96cf706676a9861f00bbab20&&language=en-US&page=1')
```

```python
df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
# Now, we have to do this work in loop. Loop to run as many pages we have

df = pd.DataFrame()
df
```

```python
# 500 are No of pages (as per API limit)

all_dfs = [] # Initialize an empty list to store dataframes
for i in range(1,501):
  response = requests.get('http://api.themoviedb.org/3/movie/top_rated?api_key=ab9e733b96cf706676a9861f00bbab20&&language=en-US&page={}'.format(i))
  json_response = response.json()
  if 'results' in json_response:
    temp_df = pd.DataFrame(json_response['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    all_dfs.append(temp_df) # Append each temporary dataframe to the list
  else:
    print(f"Warning: 'results' key not found in response for page {i}. Full response: {json_response}")
    # Optionally, you could still break here if you want to stop on any unexpected response, not just 'invalid page'
    # break

df = pd.concat(all_dfs, ignore_index=True) # Concatenate all dataframes once after the loop
```

```python
df.to_csv('movies.csv')
```
