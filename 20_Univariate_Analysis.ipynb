```python
# Using Titanic Dataset

import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
```

```python
# Step 1: Importing Data File

df = pd.read_csv('/content/Titanic_train.csv')
```

```python
df.shape
```

```python
df.sample(5)
```

```python
# Analysing Categorical Data
```

```python
# Count Plot

sns.countplot(df['Survived'])
plt.yticks(np.arange(0, 500, 100))
```

```python
df.Survived.plot(kind = 'bar')
plt.yticks(np.arange(0, 500, 100))
plt.grid()
plt.show()
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(df['Pclass'])
```

```python
df['Pclass'].value_counts().plot(kind='bar')
```

```python
df['Sex'].value_counts().plot(kind='bar')
```

```python
df['Embarked'].value_counts().plot(kind='bar')
```

```python
# Piechart

df['Sex'].value_counts().plot(kind='pie')
```

```python
import plotly.express as px
df['Pclass'].value_counts(normalize=True).plot(kind='pie', autopct='%1.1f%%')
```

```python
df['Sex'].value_counts(normalize=True).plot(kind='pie', autopct='%1.1f%%')
```

```python
# Analysing Numerical Data
```

```python
# Step 1: Histogram

plt.hist(df['Age'], bins=50)
```

```python
# Step 2: Dist Plot
# KDE -> Kernel Density Estimation
# PDF -> Probability Density Function

sns.distplot(df['Age'])
```

```python
# Step 3: Box Plot
# Give 5 Number Summary
# Median, Percentile Value, ...
# Can be used to segreggate potential outliers

sns.boxplot(df['Fare'])
```

```python
sns.boxplot(df['Age'],  orient='h')
```
